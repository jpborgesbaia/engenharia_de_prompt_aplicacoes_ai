```python
# Solicita os números ao usuário
num1_input = input("Digite o primeiro número: ")
num2_input = input("Digite o segundo número: ")

# Tenta converter os inputs para float
try:
    num1 = float(num1_input)
    num2 = float(num2_input)
except ValueError:
    print("Entrada inválida. Por favor, digite apenas números.")
else:
    # Solicita a operação ao usuário
    operation = input("Escolha a operação (+, -, *, /): ")

    result = None
    if operation == '+':
        result = add(num1, num2)
    elif operation == '-':
        result = subtract(num1, num2)
    elif operation == '*':
        result = multiply(num1, num2)
    elif operation == '/':
        result = divide(num1, num2)
    else:
        print("Operação inválida. Por favor, escolha entre +, -, * ou /.")

    if result is not None:
        print(f"O resultado da operação {num1} {operation} {num2} é: {result}")
```

```python
import psutil

def check_temperature(threshold):
    """
    Verifica a temperatura do sistema e exibe um alerta se estiver acima do limite.

    Args:
        threshold (int): O limite de temperatura em graus Celsius. Se qualquer sensor
                         reportar uma temperatura acima deste limite, um alerta será emitido.
    """
    # Informa ao usuário que a verificação de temperatura está sendo iniciada.
    print("Verificando a temperatura do sistema com limite de " + str(threshold) + "°C...")

    try:
        # Tenta obter os dados de temperatura dos sensores do sistema usando psutil.
        # psutil.sensors_temperatures() retorna um dicionário onde as chaves são nomes
        # de dispositivos (como 'cpu_thermal', 'nvme', etc.) e os valores são listas
        # de objetos 'shwtemp' contendo 'label', 'current', 'high', 'critical'.
        temps = psutil.sensors_temperatures()

        # Verifica se alguma informação de temperatura foi obtida.
        # Em alguns ambientes (como máquinas virtuais ou certos sistemas operacionais),
        # psutil pode não ser capaz de ler os sensores de temperatura.
        if not temps:
            print("Não foi possível obter dados de temperatura do sistema.")
            return # Sai da função se não houver dados.

        overheating = False # Flag para indicar se algum componente está superaquecendo.

        # Itera sobre cada grupo de sensores de temperatura.
        # 'name' pode ser, por exemplo, 'cpu_thermal', 'nvme', etc.
        # 'entries' é uma lista de objetos de temperatura para aquele grupo.
        for name, entries in temps.items():
            # Itera sobre cada entrada de temperatura dentro de um grupo de sensores.
            for entry in entries:
                # Constrói a string de saída, usando o rótulo do sensor se disponível,
                # caso contrário, usa o nome do grupo do sensor.
                sensor_name = entry.label if entry.label else name
                print("  " + sensor_name + ": " + str(entry.current) + "°C")

                # Verifica se a temperatura atual do sensor excede o limite definido.
                if entry.current > threshold:
                    # Se exceder, imprime uma mensagem de alerta de superaquecimento.
                    print("    *** ALERTA: " + sensor_name + " está superaquecendo! (" + str(entry.current) + "°C > " + str(threshold) + "°C) ***")
                    overheating = True # Define a flag de superaquecimento como True.

        # Após verificar todos os sensores, se nenhum superaquecimento foi detectado,
        # informa ao usuário que todas as temperaturas estão dentro do limite.
        if not overheating:
            print("Todas as temperaturas estão abaixo de " + str(threshold) + "°C.")

    except AttributeError:
        # Captura um AttributeError se psutil.sensors_temperatures() não for encontrado
        # ou se o sistema não suportar a funcionalidade de sensores de temperatura.
        print("Não foi possível obter informações de temperatura. O psutil pode não suportar seu sistema ou plataforma.")
    except Exception as e:
        # Captura quaisquer outras exceções inesperadas que possam ocorrer.
        print("Ocorreu um erro ao verificar a temperatura: " + str(e))

# Solicita a temperatura limite ao usuário
threshold_input = input("Digite a temperatura limite para o alerta (em °C, ex: 50): ")

try:
    # Converte a entrada do usuário para um número inteiro
    user_threshold = int(threshold_input)
    # Executa a função de verificação de temperatura com o limite fornecido pelo usuário.
    check_temperature(user_threshold)
except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro para a temperatura limite.")
```
